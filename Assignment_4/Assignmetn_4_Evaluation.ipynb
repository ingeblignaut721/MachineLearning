```python
#Not my code: https://github.com/KianAnd19/rnn-time-series/tree/main

import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
from itertools import combinations
from scipy import stats
import scikit_posthocs as sp
import pandas as pd


# https://github.com/yesteryearer/Automated-Analysis-of-Metaheuristics
def plot_critical_difference_diagram(ranks, sig_matrix):
    fig, ax = plt.subplots(figsize=(11.5, 5))
    
    sp.critical_difference_diagram(
        ranks, 
        sig_matrix, 
        ax=ax, 
        label_fmt_left='{label} [{rank:.3f}]  ',
        label_fmt_right='  [{rank:.3f}] {label}',
        text_h_margin=0.15,
        label_props={ 'fontweight': 'bold', 'fontsize': 10},
        crossbar_props={'color': None, 'marker': 'o', 'linewidth': 1.5},
        elbow_props={'color': 'gray', 'linestyle': '--'},
    )

    plt.subplots_adjust(left=0.1, right=0.9, top=0.8, bottom=0.3)

def friedman_test(data):
    k = data.shape[1]  # number of models
    n = data.shape[0]  # number of datasets
    
    # Convert data to ranks
    ranks = np.zeros_like(data)
    for i in range(n):
        ranks[i] = stats.rankdata(data[i])
    
    # Calculate R_j (sum of ranks for each model)
    R_j = np.sum(ranks, axis=0)
    
    # Calculate Friedman statistic
    chi2 = (12 * n) / (k * (k + 1)) * (np.sum(R_j ** 2) - (k * (k + 1) ** 2) / 4)
    
    # Degrees of freedom
    df = k - 1
    
    # Calculate p-value
    p_value = stats.chi2.sf(chi2, df)
    
    return chi2, p_value, df, R_j

def compute_CD(n_datasets, n_models, alpha=0.05):
    q = stats.studentized_range.ppf(q=1-alpha, k=n_models, df=np.inf)
    CD = q * np.sqrt((n_models * (n_models + 1)) / (6 * n_datasets))
    return CD

def pairwise_comparison(R_j, n_datasets, n_models):
    se = np.sqrt((n_models * (n_models + 1)) / (6 * n_datasets))
    
    comparisons = []
    for (i, Ri), (j, Rj) in combinations(enumerate(R_j), 2):
        z = (Ri - Rj) / (np.sqrt(2) * se)
        p_unadj = 2 * (1 - stats.norm.cdf(abs(z)))
        comparisons.append((i, j, z, p_unadj))
    
    # Manual Bonferroni correction
    n_comparisons = len(comparisons)
    p_adj = [min(p * n_comparisons, 1.0) for _, _, _, p in comparisons]
    
    return [(c[0], c[1], c[2], c[3], p) for c, p in zip(comparisons, p_adj)]


# Load data
data = pd.read_csv("IF_stroke.csv") 

# Drop unwanted columns
data = data.drop(columns=[col for col in data.columns if ('std' in col) or ('accuracy' in col) ])

# Combine hyperparameters to create a unique model identifier
data['model_id'] = data[['n_estimators', 'max_samples', 'contamination', 
                         'max_features', 'bootstrap']].astype(str).agg('-'.join, axis=1)

model_id = data['model_id']

# Metrics columns
metric_cols = [col for col in data.columns if col not in ['n_estimators', 'max_samples', 
                                                          'contamination', 'max_features', 
                                                          'bootstrap', 'model_id']]

# Keep only model_id + metrics
df_metrics = data[['model_id'] + metric_cols]

# Pivot: model_id becomes columns, metrics stay as rows
data_T = df_metrics.set_index('model_id').T
data_np = data_T.to_numpy()  # Transpose so metrics are rows

# Perform Friedman test
chi2, p_value, df, R_j = friedman_test(data_np)

print(f"Friedman test results:")
print(f"Chi-square statistic: {chi2:.4f}")
print(f"p-value: {p_value:.4f}")
print(f"Degrees of freedom: {df}")

if p_value < 0.05:
    print("There are significant differences among the models.")
else:
    print("There are no significant differences among the models.")

# Prepare data for critical difference plot
n_datasets, n_models = data_np.shape
average_ranks = R_j / n_datasets


top_indices = np.argsort(average_ranks)[-7:]  # top 5

# Get the corresponding model_ids
top_model_ids = [model_id[i] for i in top_indices]

# Filter the original DataFrame for these model_ids
top_models_df = data[data['model_id'].isin(top_model_ids)]

print("Top 5 models (full observations):")
display(top_models_df)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Use IEEE-like style
plt.style.use('seaborn-v0_8-whitegrid')
plt.rcParams.update({
    'font.family': 'serif',
    'font.size': 14,
    'axes.labelsize': 16,
    'axes.titlesize': 18,
    'legend.fontsize': 13,
    'xtick.labelsize': 13,
    'ytick.labelsize': 13,
    'lines.linewidth': 2,
    'lines.markersize': 7
})

# Filter and sort data
df_filtered_fraud = fraud[
    (fraud['contamination'] == 0.001) &
    (fraud['bootstrap'] == False) &
    (fraud['max_samples'] == 200) &
    (fraud['max_features'] == 0.3)
].sort_values(by='n_estimators')

df_filtered_stroke = stroke[
    (stroke['contamination'] == 0.1) &
    (stroke['bootstrap'] == False) &
    (stroke['max_samples'] == 100) &
    (stroke['max_features'] == 0.8)
].sort_values(by='n_estimators')[:-1]

df_filtered_mulcross = mulcross[
    (mulcross['contamination'] == 0.1) &
    (mulcross['bootstrap'] == False) &
    (mulcross['max_samples'] == 100) &
    (mulcross['max_features'] == 0.8)
].sort_values(by='n_estimators')

# Create figure
fig, ax = plt.subplots(figsize=(8,5))

# Plot black & white with different line styles
ax.plot(df_filtered_fraud['n_estimators'], df_filtered_fraud['train_f1_mean'],
        color='black', linestyle='-', marker='o', label='Training Fraud')
ax.plot(df_filtered_fraud['n_estimators'], df_filtered_fraud['test_f1_mean'],
        color='black', linestyle='--', marker='s', label='Testing Fraud')

ax.plot(df_filtered_stroke['n_estimators'], df_filtered_stroke['train_f1_mean'],
        color='black', linestyle='-', marker='d', label='Training Stroke')
ax.plot(df_filtered_stroke['n_estimators'], df_filtered_stroke['test_f1_mean'],
        color='black', linestyle='--', marker='v', label='Testing Stroke')

ax.plot(df_filtered_mulcross['n_estimators'], df_filtered_mulcross['train_f1_mean'],
        color='black', linestyle='-', marker='p', label='Training Mulcross')
ax.plot(df_filtered_mulcross['n_estimators'], df_filtered_mulcross['test_f1_mean'],
        color='black', linestyle='--', marker='*', label='Testing Mulcross')

# Labels and title
ax.set_xlabel('Number of Trees')
ax.set_ylabel('F1 Score')
ax.legend(frameon=False)
ax.grid(True, which='major', linestyle='--', linewidth=0.7, alpha=0.8)

# Optimize layout for publication
plt.tight_layout()
plt.show()
```
